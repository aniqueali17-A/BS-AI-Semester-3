```python
student = {
    "name": "Ali",
    "programme": "BS_AI",
    "Age": 20
}
print("keys: ", student.keys())
print("Values: ", student.values())
print("Items: ", student.items())
print("Age: ", student.get("Age"))
```

```python
student['semester'] = "3rd"
print(student)
```

```python
#removing any item using pop() and del
```

```python
student= {
    "name": "Ali",
    "Age": 20,
    "programme": "BS-AI"
}
removed = student.pop("Age")
print("Removed value: ", removed)
print("successfully deleted", student)
```

Looping using key value pair


```python
student= {
    "name": "Ali",
    "Age": 20,
    "programme": "BS-AI"
}
for s in student:
    print(student[s])
```

```python
student= {
    "name": "Ali",
    "Age": 20,
    "programme": "BS-AI"
}
for a in student.values():
    print(a)
```
